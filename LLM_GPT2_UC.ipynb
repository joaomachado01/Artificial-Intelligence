3 - Loading Tokenizer

```python
gpt2_tokenizer = GPT2Tokenizer.from_pretrained("gpt2-large")
```

```python
gpt2_tokenizer
```

4 - Loading LLM

```python
gpt2_model = GPT2LMHeadModel.from_pretrained("gpt2-large", pad_token_id=gpt2_tokenizer.eos_token_id)
```

```python
gpt2_model
```

5 - Generate text using the LLM

```python
# Prompt for LLM
prompt = "Will Agents and AI substitute people in the workforce in the future?"
```

```python
# Getting matrix ids for LLM by using tokenizer
input_ids = gpt2_tokenizer.encode(prompt, return_tensors="pt")
```

```python
# Matrix ids
input_ids
```

```python
# Generate text by passing matrix ids as input to the LLM
text_generation_output = gpt2_model.generate(
    input_ids,
    max_length = 200,
    num_beams = 5,
    no_repeat_ngram_size = 2,
    early_stopping = True
)
```

```python
# Text generation output encoded
text_generation_output
```

6 - Decoding LLM result and answer the question

```python
# Text generation output decoded
print(gpt2_tokenizer.decode(text_generation_output[0], skip_special_tokens=True))
```
